```python
run=json.loads(Path('/kaggle/working/persona_lora_run/run.json').read_text())
assert run['status']=='completed' and run['frozen_base_sha256_before']==run['frozen_base_sha256_after']
for kind in ['best','final']:
    directory=Path('/kaggle/working/persona_lora_run')/(kind+'_adapter')
    for name,meta in run['checkpoints'][kind].items():
        path=directory/name
        assert path.stat().st_size==meta['bytes'] and hashlib.sha256(path.read_bytes()).hexdigest()==meta['sha256']
for kind in ['baseline','best_adapter']:
    command=[PYTHON,str(ROOT/'generate_persona.py'),'--bank',str(ROOT/'persona_eval_40.json'),'--output','/kaggle/working/persona_evaluation/'+kind,'--revision',REVISION]
    if kind=='best_adapter':command+=['--adapter','/kaggle/working/persona_lora_run/best_adapter']
    subprocess.run(command,check=True)
# Blind pairs have anonymous answers; model mapping is stored separately.
import random
base=json.loads(Path('/kaggle/working/persona_evaluation/baseline/outputs.json').read_text())
adapted=json.loads(Path('/kaggle/working/persona_evaluation/best_adapter/outputs.json').read_text())
assert len(base)==len(adapted)==40
rng=random.Random(42)
pairs=[]
mapping=[]
for a,b in zip(base,adapted):
    assert a['id']==b['id'] and a['instruction']==b['instruction'] and a['category']==b['category']
    swap=rng.choice([False,True])
    pairs.append({'id':a['id'],'instruction':a['instruction'],'response_a':b['output'] if swap else a['output'],'response_b':a['output'] if swap else b['output']})
    mapping.append({'id':a['id'],'category':a['category'],'a':'lora' if swap else 'baseline','b':'baseline' if swap else 'lora'})
result=Path('/kaggle/working/persona_evaluation')
(result/'blind_pairs.jsonl').write_text(''.join(json.dumps(row,ensure_ascii=False)+'\n' for row in pairs))
(result/'hidden_mapping.json').write_text(json.dumps(mapping,indent=2)+'\n')
(result/'completed.json').write_text(json.dumps({'status':'completed','answers':80,'blind_pairs':40,'training_run':'persona_lora_run','revision':REVISION},indent=2)+'\n')
print('PERSONA EVALUATION EXPORT COMPLETED: 80 answers, 40 blinded pairs',flush=True)
```
